# Conversion des images au format Telmi (JPG 640x480)

Redimensionne et rogne au centre sans déformation. Version script en ligne de commande : `scripts/convert_to_telmi.py`.

```python
import os
from PIL import Image, ImageOps

# CONFIGURATION
# ---------------------------------------------------------
DOSSIER_SOURCE = "input"        # Le dossier où sont vos images
DOSSIER_SORTIE = "telmi_ready"  # Le dossier qui sera créé
TARGET_SIZE = (640, 480)        # Résolution de la Telmi
EXTENSIONS = ('.jpg', '.jpeg', '.png', '.webp', '.bmp')
# ---------------------------------------------------------

os.makedirs(DOSSIER_SORTIE, exist_ok=True)

print(f"--- Démarrage du traitement vers {TARGET_SIZE} ---")

compteur = 0

for fichier in sorted(os.listdir(DOSSIER_SOURCE)):
    if not fichier.lower().endswith(EXTENSIONS):
        continue
    try:
        chemin_entree = os.path.join(DOSSIER_SOURCE, fichier)

        with Image.open(chemin_entree) as img:
            # Tout mode non-RGB (transparence, palette, niveaux de gris...)
            # doit passer en RGB pour la sauvegarde en JPG
            if img.mode != "RGB":
                img = img.convert("RGB")

            # ImageOps.fit : redimensionne et rogne au centre (center crop)
            # en gardant les proportions, sans déformer l'image
            new_img = ImageOps.fit(img, TARGET_SIZE, method=Image.Resampling.LANCZOS, centering=(0.5, 0.5))

            # Sauvegarde en JPG pour la compatibilité maximale
            nom_base = os.path.splitext(fichier)[0]
            chemin_sortie = os.path.join(DOSSIER_SORTIE, nom_base + ".jpg")
            new_img.save(chemin_sortie, "JPEG", quality=90)

            print(f"✅ OK : {fichier}")
            compteur += 1

    except Exception as e:
        print(f"❌ Erreur sur {fichier}: {e}")

print(f"--- Terminé ! {compteur} images traitées dans le dossier '{DOSSIER_SORTIE}' ---")
```
